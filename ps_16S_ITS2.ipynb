## 2. create tax_table for ITS2

```r
## Make taxonomy table for ITS2
ITS2taxtab <- data.frame("Kingdom" = rep("Eukaryota",
                                         ncol(all3bITS2)),
                         "Phylum" = rep("Dinoflagellata",
                                        ncol(all3bITS2)),
                         "Class" = rep("Dinophyceae",
                                       ncol(all3bITS2)),
                         "Order" = rep("Suessiales",
                                       ncol(all3bITS2)),
                         "Family" = rep("Symbiodiniaceae",
                                        ncol(all3bITS2)),
                         "Genus" = NA,
                         "Species" = NA)
```

```r
rownames(ITS2taxtab) <- colnames(all3bITS2)
```

```r
for (i in 1:nrow(ITS2taxtab)) {
  if (startsWith(rownames(ITS2taxtab)[i], "A") == TRUE) {
    ITS2taxtab$Genus[i] <- "Symbiodinium"
  }
  else if (endsWith(rownames(ITS2taxtab)[i], "A") == TRUE) {
    ITS2taxtab$Genus[i] <- "Symbiodinium"
  }
  else if (startsWith(rownames(ITS2taxtab)[i], "B") == TRUE) {
    ITS2taxtab$Genus[i] <- "Breviolum"
  }
  else if (endsWith(rownames(ITS2taxtab)[i], "B") == TRUE) {
    ITS2taxtab$Genus[i] <- "Breviolum"
  }
  else if (startsWith(rownames(ITS2taxtab)[i], "C") == TRUE) {
    ITS2taxtab$Genus[i] <- "Cladocopium"
  }
  else if (endsWith(rownames(ITS2taxtab)[i], "C") == TRUE) {
    ITS2taxtab$Genus[i] <- "Cladocopium"
  }
  else if (startsWith(rownames(ITS2taxtab)[i], "D") == TRUE) {
    ITS2taxtab$Genus[i] <- "Durusdinium"
  }
  else if (endsWith(rownames(ITS2taxtab)[i], "D") == TRUE) {
    ITS2taxtab$Genus[i] <- "Durusdinium"
  }
  else {
    ITS2taxtab$Genus[i] <- "NA"
  }
}
```

```r
# Make joint taxonomy table
taxtab16S <- data.frame(tax_table(all3b16S))
taxtabboth <- rbind(ITS2taxtab,
                    taxtab16S)
```
